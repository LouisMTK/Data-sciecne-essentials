```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

if 'student_df' not in globals():
    data = [50, 50, 47, 97, 49, 3, 53, 42, 26, 74, 82, 62, 37, 15, 70, 27, 36, 35, 48, 52, 63, 64]
    grades = np.array(data)
    study_hours = np.array([2, 3, 1, 5, 4, 1, 2, 3, 2, 4, 5, 3, 2, 1, 4, 2, 3, 2, 4, 5, 3, 2])
    names = ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Henry', 'Ivy', 'Jack', 'Kate', 'Liam', 'Mia', 'Noah', 'Olivia', 'Paul', 'Quinn', 'Ruby', 'Sam', 'Tara', 'Uma', 'Victor']
    student_df = pd.DataFrame({'name': names, 'study_hours': study_hours, 'grades': grades})

if 'passed' not in student_df.columns:
    student_df['passed'] = student_df['grades'] >= 60

pass_counts = student_df['passed'].value_counts().reindex([False, True], fill_value=0)

fig, ax = plt.subplots(1, 2, figsize=(12.8, 6.5))
ax[0].bar(student_df['name'], student_df['grades'], color='green')
ax[0].set_title('Student Grades')
ax[0].set_xlabel('Students')
ax[0].set_ylabel('Grades')
ax[0].tick_params(axis='x', rotation=90)

ax[1].pie(pass_counts.values, labels=['Fail', 'Pass'], autopct='%1.1f%%')
ax[1].set_title('Passing Grades')

plt.tight_layout()
plt.show()
```
